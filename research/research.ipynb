```python
## Ensure annotations

def get_product(x:int,y:int)-> int:
    return x*y
get_product(5,2)
```

```python
get_product(4,"5")
```

```python
from ensure import ensure_annotations
@ensure_annotations
def get_product(x:int,y:int)-> int:
    return x*y
get_product(5,2)
```

```python
get_product(4,"5")
```
